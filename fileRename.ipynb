```python
import os
import pandas as pd

# Excel文件路径和工作表名称
excel_path ='C:/Users/Administrator/Desktop/重新补样信息new.xlsx'
sheet_name = 'Sheet1'  # Excel工作表名称，根据实际情况调整

# 文件夹路径，其中包含需要重命名的文件
folder_path = 'C:/Users/Administrator/Desktop/0129提取处理'

# 读取Excel文件
df = pd.read_excel(excel_path, sheet_name=sheet_name)

# 遍历所有行
for index, row in df.iterrows():
    name2_base = row['name']  # 根据实际列名调整
    name1_base = row['rawname']  # 根据实际列名调整
    
    # 在指定文件夹中搜索与名称2匹配的文件
    for filename in os.listdir(folder_path):
        # 分离文件名和扩展名
        base_name, ext = os.path.splitext(filename)
        if base_name == name2_base:
            original_file = os.path.join(folder_path, filename)
            new_file = os.path.join(folder_path, f"{name1_base}{ext}")

            # 重命名文件
            os.rename(original_file, new_file)
            print(f'Renamed "{filename}" to "{name1_base}{ext}"')
            break  # 假设每个名称2在文件夹中只对应一个文件
    else:
        print(f'File matching "{name2_base}" not found.')

print('Renaming completed.')
```
