```python
import re

# 读取 Excel 文件
df = pd.read_excel('C:/Users/Administrator/Desktop/ANALYSIS.xlsx', sheet_name='PB MSMS')

# 提取脂质特征的列
lipid_column = df['lipids'].dropna()

# 定义一个函数来标准化脂质特征的格式并返回排序后的元组
def normalize_lipid_feature(lipid):
    # 使用正则表达式匹配_前后的部分
    matches = re.findall(r'(\d+):(\d+)', lipid)
    # 返回排序后的元组
    return tuple(sorted(matches))

# 对每个脂质特征进行标准化处理
normalized_lipids = lipid_column.apply(normalize_lipid_feature)
normalized_lipids=normalized_lipids.to_numpy()

print(np.shape(normalized_lipids))
```

```python
# 初始化一个字典，用于存储每个 unique_lipids 元素的计数
lipid_counts = {}

# 遍历 unique_lipids 中的每个元素
for unique_lipid in unique_lipids:

    # 初始化计数器
    count = 0
    # 遍历 normalized_lipids 中的每个元素
    for lipid in normalized_lipids:

        # 如果当前元素与 unique_lipid 相等，则增加计数
        if lipid==unique_lipid:
            count += 1
    # 将计数结果存储在字典中
    lipid_counts[unique_lipid] = count

print(lipid_counts)
```

```python
# # 初始化总计数为0
total_count = 0
lipid_counts.values()
# 遍历计数字典中的每个键值对，将计数值相加
for count in lipid_counts.values():
    total_count += count

print(total_count)
```
