```python
import pandas as pd
import os
import shutil

# Excel文件路径
excel_path = 'E:/神经胶质瘤/HS_data_new/labellist.xlsx'

# 源文件夹路径
source_folder = 'E:/神经胶质瘤/HS_data_new'

# 目标文件夹路径
target_folder = 'E:/神经胶质瘤/HS_data_new_CWT'

# 读取Excel文件
df = pd.read_excel(excel_path)

print(df.iloc[:, 0])

# 遍历DataFrame
for index, row in df.iterrows():
    folder_name = str(row[0])  # 第一列数据作为文件夹名称
    condition = row[1]  # 第二列数据作为条件

    # 检查条件是否为0或1
    if condition in [0, 1]:
        source_dir = os.path.join(source_folder, folder_name)
        target_dir = os.path.join(target_folder, folder_name)

        # 检查源目录是否存在且为文件夹
        if os.path.exists(source_dir) and os.path.isdir(source_dir):
            # 如果目标目录不存在，则创建
            os.makedirs(target_dir, exist_ok=True)

            # 复制整个文件夹
            shutil.copytree(source_dir, target_dir, dirs_exist_ok=True)
```
